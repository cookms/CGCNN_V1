```python
import pandas as pd
from pathlib import Path
import os

df = pd.read_csv("id_prop.csv")


# Shuffle once with a fixed seed.
df = df.sample(frac=1, random_state=0).reset_index(drop=True)

df["cif_path"]=df['cif_path'].str.replace("materials_gnn/data/cifs_all/", "../cifs_all/", regex=False)


Path("data/subsets").mkdir(parents=True, exist_ok=True)

df.iloc[:32].to_csv("data/subsets/id_prop_32.csv", index=False)
df.iloc[:256].to_csv("data/subsets/id_prop_256.csv", index=False)
df.iloc[:1000].to_csv("data/subsets/id_prop_1000.csv", index=False)
df.iloc[:10000].to_csv("data/subsets/id_prop_10000.csv", index=False)
df.iloc[:30000].to_csv("data/subsets/id_prop_30000.csv", index=False)
df.iloc[:50000].to_csv("data/subsets/id_prop_50000.csv", index=False)
```
